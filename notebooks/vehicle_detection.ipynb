```python
import cv2
import os
from datetime import datetime
import glob
```

```python
DATA_FOLDER = "data/videos"                 # ← FIXED: Now points to videos subfolder
OUTPUT_BASE = "data/extracted_frames"       # Frames will be saved here

# Supported video extensions
VIDEO_EXTENSIONS = ["*.mp4", "*.avi", "*.mov", "*.mkv"]

FRAME_INTERVAL = 1                          # Change to 2, 3, etc. if you want fewer frames
MAX_FRAMES_PER_VIDEO = None                 # None = all frames, or set 500 to limit

os.makedirs(OUTPUT_BASE, exist_ok=True)

print("🚀 Starting Data Input Module...")
print(f"📁 Looking for videos in: {DATA_FOLDER}")
print(f"📁 Frames will be saved in: {OUTPUT_BASE}\n")
```

```python
video_files = []
for ext in VIDEO_EXTENSIONS:
    video_files.extend(glob.glob(os.path.join(DATA_FOLDER, ext)))

if not video_files:
    print("❌ Still no videos found! Please check:")
    print("   1. Are the videos really inside data/videos/ ?")
    print("   2. File names end with .mp4 ?")
    print("   3. Run the script from the project root folder")
    exit()

print(f"✅ Found {len(video_files)} video(s):")
for i, video in enumerate(video_files, 1):
    print(f"   {i}. {os.path.basename(video)}")
print()
```
